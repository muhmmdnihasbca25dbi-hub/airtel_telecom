```python
import sqlite3                                 
import pandas as pd                             
 
con = sqlite3.connect('../data/raw/airtel.db')  
 
def q(sql):                                     
    return pd.read_sql(sql, con)
```

```python
def run_sql_file(file_path):
    with open(file_path, 'r') as file:
        sql_script = file.read()
        con.executescript(sql_script)
        con.commit()
        print("ran", file_path)
```
